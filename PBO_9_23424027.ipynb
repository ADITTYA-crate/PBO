```python
# IMPORT DATACLASS
from dataclasses import dataclass

@dataclass
class Dosen:
    nama: str  # atribut nama dosen


@dataclass
class Mahasiswa:
    nama: str  # atribut nama mahasiswa

    def belajar(self, dosen):
        # method ini menunjukkan association
        # mahasiswa menggunakan objek dosen
        print(f"{self.nama} belajar dengan dosen {dosen.nama}")

@dataclass
class MahasiswaAgg:
    nama: str  # atribut mahasiswa


@dataclass
class Universitas:
    nama: str

    def __init__(self, nama, mahasiswa_list):
        self.nama = nama
        # universitas "punya" mahasiswa (aggregation)
        # tapi mahasiswa bisa hidup tanpa universitas
        self.mahasiswa_list = mahasiswa_list

    def tampilkan_mahasiswa(self):
        print(f"Daftar mahasiswa di {self.nama}:")
        for m in self.mahasiswa_list:
            print("-", m.nama)

@dataclass
class Mesin:
    def nyala(self):
        print("Mesin menyala")


@dataclass
class Mobil:
    def __init__(self, nama):
        self.nama = nama
        # composition: mesin dibuat di dalam mobil
        # mesin tidak bisa ada tanpa mobil
        self.mesin = Mesin()

    def jalan(self):
        print(f"{self.nama} berjalan")
        self.mesin.nyala()

print("=== ASSOCIATION ===")
d1 = Dosen("Pak Budi")
m1 = Mahasiswa("Ani")

m1.belajar(d1)  # mahasiswa pakai dosen


print("\n=== AGGREGATION ===")
mhs1 = MahasiswaAgg("Budi")
mhs2 = MahasiswaAgg("Siti")

# mahasiswa dibuat di luar universitas
kampus = Universitas("UNIV A", [mhs1, mhs2])

kampus.tampilkan_mahasiswa()


print("\n=== COMPOSITION ===")
mobil = Mobil("Toyota")
mobil.jalan()
```
